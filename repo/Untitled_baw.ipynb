```python
import os, sys

# ── re-establish working directory after session restart ──
os.chdir('/content/biochemical-a.w')
sys.path.insert(0, '/content/biochemical-a.w')   # ensures baw_core is findable

from baw_core.parser   import ChemistryMusicParser
from baw_core.embedder import BiochemicalEmbedder
from baw_core.detector import BiochemicalTransformerDetector

parser   = ChemistryMusicParser()
embedder = BiochemicalEmbedder()
detector = BiochemicalTransformerDetector()

AUDIO_IN  = "baw_core/your_audio (1).wav"
AUDIO_OUT = "watermarked.wav"

packet = parser.build_packet(AUDIO_IN, "C8H10N4O2", sequence_length=50)
embedder.embed_signature(AUDIO_IN, AUDIO_OUT, packet)
print("✅ watermarked.wav created")

score = detector.verify_audio_integrity(AUDIO_OUT, packet)
print(f"Integrity score: {score:.4f}")
print("→ Intact" if score > 0.5 else "→ Tampered / cloned")
```

```python
# simulate a "clone attack" — aggressive compression destroys fine-grained signal
import librosa, soundfile as sf
import numpy as np

# load watermarked audio and brutally re-encode it (lossy roundtrip)
audio, sr = librosa.load(AUDIO_OUT, sr=None, mono=True)
# simulate source-separation attack: zero out top-half frequencies
stft = librosa.stft(audio)
stft[stft.shape[0]//2:, :] = 0          # wipe the band where signature lives
attacked = librosa.istft(stft)
sf.write("attacked.wav", attacked, sr)

# now check if the detector catches it
score_attacked = detector.verify_audio_integrity("attacked.wav", packet)
print(f"Original score:  {score:.4f}  → {'Intact' if score > 0.5 else 'Tampered'}")
print(f"Attacked score:  {score_attacked:.4f}  → {'Intact' if score_attacked > 0.5 else 'Tampered'}")
```
